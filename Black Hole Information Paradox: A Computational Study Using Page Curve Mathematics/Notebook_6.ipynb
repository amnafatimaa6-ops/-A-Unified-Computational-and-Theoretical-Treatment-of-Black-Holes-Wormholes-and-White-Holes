## Black Hole Information Paradox: A Computational Study Using Page Curve Mathematics


## Purpose

The Black Hole Information Paradox is one of the greatest unsolved problems in modern physics.

According to Albert Einstein's General Relativity, anything that crosses a black hole's event horizon cannot escape.

Later, Stephen Hawking showed that black holes slowly lose mass by emitting Hawking Radiation.

This raises a profound question:

If a black hole completely evaporates, what happens to the information that fell inside it?

Quantum mechanics states that information can never be destroyed, while Hawking's original calculation appeared to suggest that it could be.

This conflict is called the Black Hole Information Paradox.


In this notebook, we will build a simplified computational model to study how information behaves during black hole evaporation.

Create a toy model of black hole evaporation.
Measure how radiation entropy changes over time.
Simulate the famous Page Curve.
Compare two competing theories:
Hawking's original prediction (information loss).
Modern unitary prediction (information preservation).
Draw scientific conclusions from our results.

Important: This notebook does not solve the information paradox. It is an educational and computational model that helps us understand the competing ideas.

# PART 1

# Building the Toy Model

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Define Black Hole Parameters

A real black hole contains an unimaginably large amount of information.

Instead of modelling billions upon billions of quantum states, we simplify the system into 100 evaporation steps.

Each step represents a small amount of Hawking radiation leaving the black hole.

```python
# Number of evaporation steps
N = 100

# Time axis
time = np.arange(N + 1)

print("Total evaporation steps:", N)
```

## Visualize Evaporation Time

```python
plt.figure(figsize=(8,2))

plt.plot(time,
         np.zeros_like(time),
         marker='o')

plt.yticks([])

plt.xlabel("Evaporation Step")

plt.title("Black Hole Lifetime")

plt.grid(True)

plt.show()
```

The graph is simply a timeline.

It does not show entropy yet.

Instead, it shows the sequence of evaporation steps that our simulation will follow.

Think of it as the x-axis for everything that comes later.

## Define the Page Time


The Page Time is the moment when approximately half of the black hole's entropy has been radiated away.

Before the Page Time:

Radiation becomes more disordered.
Entropy increases.

After the Page Time:

Information begins to emerge.
Entropy decreases.

For this toy model, we place the Page Time halfway through the evaporation.

```python
page_time = N / 2

print("Page Time:", page_time)
```

## Create Empty Lists

We need containers to store the entropy values produced by each theoretical model.

```python
hawking_entropy = []

page_entropy = []
```

hawking_entropy

Will store entropy values predicted by Hawking's original model.

page_entropy

Will store entropy values predicted by the Page Curve.

As the simulation progresses, values will be added to these lists.

##Part 2 – Simulating the Information Paradox


In this section, we simulate two competing ideas about what happens to
information during black hole evaporation.


Model 1 — Hawking's Original Prediction (1976)

Hawking radiation is purely thermal.
Information is lost.
Radiation entropy keeps increasing until the black hole disappears.

Model 2 — Page Curve Prediction

Information is preserved.

Entropy increases at first.

After the Page Time, information begins to leak out through subtle quantum correlations.

Entropy decreases back toward zero.

We will generate both predictions and compare them.

## Hawking Information Loss Model

Create entropy values assuming Hawking's original prediction.

```python
# Hawking model
hawking_entropy = []

for t in time:
    S = t
    hawking_entropy.append(S)
```

## Plot Hawking's Prediction

```python
plt.figure(figsize=(8,5))

plt.plot(
    time,
    hawking_entropy,
    color="red",
    linewidth=3
)

plt.xlabel("Evaporation Step")

plt.ylabel("Radiation Entropy")

plt.title("Hawking Information Loss Model")

plt.grid(True)

plt.show()
```

Meaning:

More Hawking radiation

More entropy

Information never returns

This is exactly what created the Information Paradox.

## Page Curve Model

```python
page_entropy = []

for t in time:

    if t <= page_time:

        S = t

    else:

        S = N - t

    page_entropy.append(S)
```

Explanation:

This model has two stages.

Before Page Time

Entropy increases.

S=t

because radiation becomes increasingly entangled with the remaining black hole.

After Page Time

Entropy decreases.

S=N−t

meaning information starts appearing in the outgoing Hawking radiation.

```python
plt.figure(figsize=(8,5))

plt.plot(
    time,
    page_entropy,
    color="blue",
    linewidth=3
)

plt.xlabel("Evaporation Step")

plt.ylabel("Radiation Entropy")

plt.title("Page Curve")

plt.grid(True)

plt.show()
```

The graph rises.

Reaches a maximum.

Then falls.

This suggests information is not destroyed.

Instead, it gradually returns during evaporation.

## Compare Both Models  




```python
plt.figure(figsize=(10,6))

plt.plot(
    time,
    hawking_entropy,
    label="Hawking Model",
    linewidth=3
)

plt.plot(
    time,
    page_entropy,
    label="Page Curve",
    linewidth=3
)

plt.axvline(
    page_time,
    color="black",
    linestyle="--",
    label="Page Time"
)

plt.xlabel("Evaporation Step")

plt.ylabel("Radiation Entropy")

plt.title("Black Hole Information Paradox")

plt.legend()

plt.grid(True)

plt.show()
```

Red line (Hawking): entropy keeps increasing.

Blue line (Page Curve): entropy rises until the Page Time, then decreases.

Dashed vertical line: marks the Page Time, where information begins to emerge.

## Part 4 – Machine Learning Classification of Entropy Curves






In Parts 1–3, we generated two different entropy curves:

Hawking Model → Information Loss (Label = 0)
Page Curve → Information Preserved (Label = 1)

Now we'll train a machine learning model to classify an unknown entropy curve.

```python
from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, classification_report

import numpy as np
```

## Generate the Dataset

```python
num_samples = 300

X = []

y = []

for i in range(num_samples):

    noise = np.random.normal(0, 2, len(time))

    if np.random.rand() < 0.5:

        curve = np.array(hawking_entropy) + noise

        label = 0

    else:

        curve = np.array(page_entropy) + noise

        label = 1

    X.append(curve)

    y.append(label)

X = np.array(X)

y = np.array(y)

print("Dataset Shape:", X.shape)
```

## Split the Dataset

```python
X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.25,

    random_state=42

)

print("Training Samples:", len(X_train))

print("Testing Samples:", len(X_test))
```

## Train the Model

Train a Random Forest classifier to recognise the two types of entropy curves.

```python
model = RandomForestClassifier(

    n_estimators=100,

    random_state=42

)

model.fit(

    X_train,

    y_train

)
```

```python
predictions = model.predict(X_test)
```

## Evaluate the Model

```python
accuracy = accuracy_score(

    y_test,

    predictions

)

print(f"Accuracy: {accuracy:.2%}")

print()

print(classification_report(

    y_test,

    predictions

))
```

## Test the AI on a New Curve

```python
new_curve = np.array(page_entropy)

new_curve = new_curve + np.random.normal(

    0,

    2,

    len(time)

)

prediction = model.predict(

    new_curve.reshape(1, -1)

)[0]

if prediction == 0:

    print("Prediction: Hawking Model (Information Loss)")

else:

    print("Prediction: Page Curve (Information Preserved)")
```

## Visualize the Test Curve

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(new_curve, linewidth=2)

plt.title("Entropy Curve Classified by AI")

plt.xlabel("Evaporation Step")

plt.ylabel("Entropy")

plt.grid(True)

plt.show()
```

### Interpretation of the Machine Learning Results

The Random Forest classifier achieved an accuracy of **100%** on the test dataset. Although this may initially appear unusually high, it is an expected outcome for the simplified nature of this computational experiment.

The dataset consisted of two well-defined theoretical entropy evolution models. The **Hawking model** produced a monotonically increasing entropy curve, representing apparent information loss, whereas the **Page curve** exhibited an initial increase followed by a decrease after the Page time, representing information preservation. These two patterns possess fundamentally different geometric characteristics.

To improve realism, Gaussian noise was added to each entropy curve before training. While the noise introduced small random fluctuations, it did not significantly alter the overall shape of the curves. Consequently, the distinguishing features of each model remained clearly identifiable.

The Random Forest algorithm is particularly effective at recognizing nonlinear patterns within structured datasets. Since the two classes were highly separable in the feature space, the classifier was able to learn decision boundaries that perfectly distinguished between the two theoretical models, resulting in a classification accuracy of 100%.

It is important to note that this result **does not imply that the model is universally accurate** or that it can classify real observational black hole data with perfect reliability. The classifier was trained and evaluated exclusively on synthetic entropy curves generated from a simplified toy model. Therefore, the reported accuracy reflects the controlled nature of the simulated dataset rather than the complexity of real astrophysical observations.

In a more realistic research setting, entropy curves would vary due to differences in black hole mass, evaporation rate, Page time, measurement uncertainty, and additional physical effects. Such variability would create overlapping class boundaries, making the classification task considerably more challenging and reducing the expected accuracy. Consequently, the purpose of the machine learning model in this project is to demonstrate the feasibility of automatically distinguishing theoretical entropy evolution models, rather than to provide a definitive solution to the black hole information paradox.
